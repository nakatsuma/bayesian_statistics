### Monte Carlo Method


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng(seed=9999)
max_n = 1000000
n = range(1, max_n + 1)
```

```python
x = rng.normal(size=max_n)
```

#### Population mean

$$
  \bar{X}_{n} = \frac1{n}\sum_{i=1}^{n}X_{i},\quad \mathrm{E}[X_{i}] = 0
$$

```python
bar_x = np.cumsum(x) / n
plt.plot(np.log10(n), bar_x)
plt.xlabel("$\\log_{10} n$")
plt.ylabel("$\\bar{x}_{n}$")
plt.show()
```

#### Probability

$$
  \hat{P} = \frac1{n}\sum_{i=1}^{n}\mathbb{1}(X_{i}\leqq 0),\quad \Pr\{X_{i}\leqq 0\} = 0.5,\quad   \mathbb{1}(A) = \begin{cases}
  1 & \text{if $A$ is true;} \\
  0 & \text{if $A$ ls false.}
  \end{cases}
$$


```python
hat_p = np.cumsum((x <= 0.0)) / n
plt.plot(np.log10(n), hat_p)
plt.xlabel("$\\log_{10} n$")
plt.ylabel("$\\hat{p}_{n}$")
plt.show()
```
